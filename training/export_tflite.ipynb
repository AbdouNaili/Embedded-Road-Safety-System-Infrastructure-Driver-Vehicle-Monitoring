```python
from ultralytics import YOLO
import os

# =========================================================
# 1️⃣ MODEL PATH
# =========================================================
model_path = "runs/detection_routiere_v1/weights/best.pt"
model = YOLO(model_path)

# =========================================================
# 2️⃣ DATASET YAML (for INT8 calibration)
# =========================================================
dataset_yaml = "dataset/data.yaml"

# =========================================================
# 3️⃣ OUTPUT DIRECTORY
# =========================================================
output_dir = "exports"
os.makedirs(output_dir, exist_ok=True)

# =========================================================
# 4️⃣ IMAGE SIZE
# =========================================================
IMG_SIZE = 416

# =========================================================
# 5️⃣ EXPORT ONNX
# =========================================================
print("\n➡️ Exporting ONNX...")
model.export(
    format="onnx",
    imgsz=IMG_SIZE,
    project=output_dir,
    name="onnx_416",
    exist_ok=True,
    verbose=True
)

# =========================================================
# 6️⃣ EXPORT TFLite FP32
# =========================================================
print("\n➡️ Exporting TFLite FP32...")
model.export(
    format="tflite",
    imgsz=IMG_SIZE,
    int8=False,
    half=False,
    project=output_dir,
    name="tflite_fp32_416",
    exist_ok=True,
    verbose=True
)

# =========================================================
# 7️⃣ EXPORT TFLite FP16
# =========================================================
print("\n➡️ Exporting TFLite FP16...")
model.export(
    format="tflite",
    imgsz=IMG_SIZE,
    half=True,
    int8=False,
    project=output_dir,
    name="tflite_fp16_416",
    exist_ok=True,
    verbose=True
)

# =========================================================
# 8️⃣ EXPORT TFLite INT8
#    One export writes both best_int8.tflite and
#    best_full_integer_quant.tflite (the one used on the Raspberry Pi)
# =========================================================
print("\n➡️ Exporting TFLite INT8...")
model.export(
    format="tflite",
    imgsz=IMG_SIZE,
    int8=True,
    data=dataset_yaml,
    fraction=0.1,
    project=output_dir,
    name="tflite_int8_416",
    exist_ok=True,
    verbose=True
)

# =========================================================
# DONE
# =========================================================
print("\n✅ ALL EXPORTS COMPLETED!")
print(f"📁 Saved in: {output_dir}")
```
